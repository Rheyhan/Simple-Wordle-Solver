# Basic algorithm for what will be used for the deployment

```python
import re
import json
import pandas as pd
from typing import *
```

```python
word_lists = pd.read_csv("dict/5_common.csv").values
word_lists = [str(i[0]).lower() for i in word_lists]
```

```python
# Shit baseline algorithm

letter1 = 'a'
letter2 = '.'
letter3 = '.'
letter4 = 'i'
letter5 = '.'

search = (letter1 + letter2 + letter3 + letter4 + letter5).lower()
blacklisted = 'bcd'.lower()
possible = ''.lower()


pattern = search.replace('.', '[a-z]')

filtered = [word for word in word_lists if re.match(f'^{pattern}$', word)]
if blacklisted:
    filtered = [word for word in filtered if not any(char in word for char in blacklisted)]
filtered = sorted([word for word in filtered if all(char in word for char in possible)])

# All possible words
print(filtered)
```
